Part 2

```python
def f(x):
  return x**2

def estimate_area(a, b, n):
  width = (b - a) / n
  total_area = 0

  for i in range(n):
    x = a + i * width
    height = f(x)
    rectangle_area = height * width
    total_area = total_area + rectangle_area
  return total_area

actual_area = 8/3
estimated_area = estimate_area(0, 2, 10)

print("Actual area:", actual_area)
print("Estimated area:", estimated_area)
print("Difference:", actual_area - estimated_area)
```

Part 3

```python
n_values = [10, 100, 1000, 10000, 100000]

print(f"Actual Area: {actual_area}\n")
print(f"{'n':<10} | {'Estimated Area':<20} | {'Difference':<20}")
print("-" * 58)

for n in n_values:
    est = estimate_area(0, 2, n)
    diff = actual_area - est
    print(f"{n:<10} | {est:<20.8f} | {diff:<20.8e}")
```
