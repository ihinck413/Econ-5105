Problem 1

```python
import numpy as np


A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

add = A + B
sub = A - B

twoA = 2 * A

AB = np.dot(A, B)
BA = np.dot(B, A)

print("A + B =")
print(add)
print("\nA - B =")
print(sub)
print("\n2A =")
print(twoA)
print("\nAB =")
print(AB)
print("\nBA =")
print(BA)

are_equal = np.array_equal(AB, BA)
print(f"\nAre AB and BA equal? {are_equal}")
```

Problem 3

```python
import numpy as np

A = np.array([[2, 1], [3, -1]])
b = np.array([5, 4])

A_inv = np.linalg.inv(A)

x_sol = np.dot(A_inv, b)

print("Matrix A Inverse (A^-1):")
print(A_inv)
print("\nSolution vector x (containing x and y):")
print(x_sol)
print(f"\nCalculated Solutions: x = {x_sol[0]:.4f}, y = {x_sol[1]:.4f}")
```

Problem 4

```python
import numpy as np

C = np.array([[1, 2, 3],
              [1, 1, 1],
              [3, 5, 7]])

det_C = np.linalg.det(C)

rank_C = np.linalg.matrix_rank(C)

print(f"(a) Determinant of C, |C| = {det_C:.4f}")
print(f"(b) Rank of C = {rank_C}")
```
